```python
#Créer une Series contenant 4 températures
import pandas as pd
temperatures= pd.Series([24.5, 25.1, 26.3, 27.0])
temperatures
```

```python
#Afficher son index 
print(temperatures.index)
```

```python
#Afficher la 1ère valeur 
print("la 1ère valeur est :", temperatures.iloc[0])
```

```python
# Afficher la dernière valeur 
print("la dernière valeur est :", temperatures.iloc[-1])
```

```python
#Créer, de deux façon différentes, une Series de 4 températures 
# avec comme index les heures de mesure 12h, 13h, 14h et 15h 
temperatures_heures= pd.Series([24.5, 25.1, 26.3, 27.0], index=['12h', '13h', '14h', '15h'])
print(temperatures_heures)

```

```python
temperature_dict = {'12h': 24.5, '13h': 25.1, '14h': 26.3, '15h': 27.0}
temperatures_heures_dict = pd.Series(temperature_dict)
print(temperatures_heures_dict)
```

```python
temperatures_heures2= pd.Series([24.5, 25.1, 26.3, 27.0])
temperatures_heures2.index = ['12h', '13h', '14h', '15h']
print(temperatures_heures2)
                                
```

```python
#Afficher la 1ère et la dernière valeur 
print("la 1ère valeur est :", temperatures_heures.iloc[0])
print("la dernière valeur est :", temperatures_heures.iloc[-1])
```
